```python
import numpy as np
import matplotlib.pyplot as plt
import math

def taylor_sin(x, n_terms):
    approximation = 0
    for n in range(n_terms):
        exponent = 2 * n + 1
        term = ((-1)**n * (x**exponent)) / math.factorial(exponent)
        approximation += term
    return approximation

# x values from 0 to 10 with steps of 0.1
x_values = np.arange(0, 10.1, 0.1)
exact_values = np.sin(x_values)

# approximations and absolute errors for first 4 terms
term_counts = [1, 2, 3, 4]
errors = {}

plt.figure(figsize=(10, 6))

for n in term_counts:
    approx = np.array([taylor_sin(x, n) for x in x_values])
    error = np.abs(exact_values - approx)
    errors[n] = error
    plt.plot(x_values, error, label=f'{n} Term(s)')

# 3 Plotting Truncation Error
plt.xlabel('x value')
plt.ylabel('Absolute Error')
plt.title('Truncation Error of Taylor Series Approximation for sin(x)')
plt.legend()
plt.grid(True)
plt.show()

# Citation: Taylor Series formula for sine centered at 0 (Maclaurin Series)
# Discussion: The Taylor series is very accurate near x=0. However, the error 
# grows quickly as x increases. Even using 4 terms, the approximation fails 
# to stay accurate by the time it reaches x=10. This shows that the truncation
# error is not uniform across all values of x. It stays small for low values 
# but diverges significantly as x gets larger.
```
