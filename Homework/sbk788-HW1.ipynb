```python
import math
import random
import matplotlib.pyplot as plt

# Monte Carlo Pi Function
def approxPi(numSamples):
    count = 0
    for i in range(0, numSamples):
        # Generate random coordinates
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        
        # Check if point is within circle
        if (x*x + y*y < 1):
            count += 1
            
    # Pi approximation
    return 4 * (count / numSamples)

# Iterate from 10 to 1000 in steps of 20
samples = list(range(10, 1001, 20)) 
absErrors = []
realPi = math.pi 

for i in samples:
    codePi = approxPi(i)
    absError = abs(realPi - codePi)
    absErrors.append(absError)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(samples, absErrors, 'b-o', markersize=4)
plt.title('Monte Carlo Pi Absolute Error (Up to 1000 Samples)')
plt.xlabel('Number of Samples')
plt.ylabel('Absolute Error')
plt.grid(True)
plt.tight_layout()
plt.show()

# Citation: Monte Carlo method for Pi estimation via area of a circle
```
