```python
import math
import random
import matplotlib.pyplot as plt

def approxPi(numSamples):
    count = 0
    for i in range(0, numSamples):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if (x*x + y*y < 1):
            count+=1
    return 4*(count/numSamples)

samples = list(range(10, 10001, 20))
errors = []
realPi = math.pi

for i in samples:
    codePi = approxPi(i)
    error = abs(codePi - realPi)
    errors.append(error)

plt.plot(samples, errors, 'b-o')
plt.title('Monte Carlo Pi Approximation Error')
plt.xlabel('Number of Samples')
plt.ylabel('Absolute Error')
plt.grid(True)
plt.show()
```
