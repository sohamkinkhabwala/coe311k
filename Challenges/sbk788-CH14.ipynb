```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi
N = 50

# Create uniform grid
x = np.linspace(a, b, N)

# Step size
h = (b - a) / (N - 1)

# Approximations
df_central  = (f(x + h) - f(x - h)) / (2 * h)
df_forward  = (f(x + h) - f(x)) / h
df_backward = (f(x) - f(x - h)) / h
df_exact    = np.cos(x)

plt.figure(figsize=(14, 10))
plt.suptitle("Approximating Derivative of f(x) = sin(x)", fontsize=16)

# Exact derivative plotting
plt.subplot(2, 2, 1)
plt.plot(x, df_exact, color="blue", linewidth=1.5, label="cos(x)")
plt.title("Exact Derivative: cos(x)", pad=10)
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.legend()

# Forward Difference plotting
plt.subplot(2, 2, 2)
plt.plot(x, df_exact, color="blue", linestyle="--", linewidth=1.5, label="cos(x)")
plt.plot(x, df_forward, color="orange", marker="o", markersize=3, linewidth=1.5, label="Forward Diff")
plt.title("Forward Difference", pad=10)
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.legend()

# Backward Difference plotting
plt.subplot(2, 2, 3)
plt.plot(x, df_exact, color="blue", linestyle="--", linewidth=1.5, label="cos(x)")
plt.plot(x, df_backward, color="green", marker="o", markersize=3, linewidth=1.5, label="Backward Diff")
plt.title("Backward Difference", pad=10)
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.legend()

# Central Difference plotting
plt.subplot(2, 2, 4)
plt.plot(x, df_exact, color="blue", linestyle="--", linewidth=1.5, label="cos(x)")
plt.plot(x, df_central, color="red", marker="o", markersize=3, linewidth=1.5, label="Central Diff")
plt.title("Central Difference", pad=10)
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.legend()

plt.show()
```
