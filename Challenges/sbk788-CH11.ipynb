```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

# data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

# cubine spline
cs = CubicSpline(x, y, bc_type='natural')

# new x values for smooth plot
x_interp = np.linspace(0, 5, 100)
y_interp = cs(x_interp)

# polynomial fit
degree = 5  # degree = data points-1
coeffs = np.polyfit(x, y, degree)
poly = np.poly1d(coeffs)
y_poly = poly(x_interp)

# plotting
plt.figure(figsize=(9, 5))

plt.plot(x_interp, y_interp, label='Cubic Spline', color='blue', linewidth=2)
plt.plot(x_interp, y_poly,   label=f'Polynomial (degree {degree})', color='red', linewidth=2, linestyle='--')
plt.scatter(x, y, color='black', label='Data Points')

plt.title('Cubic Spline vs Polynomial Fitting')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
