```python
import math
import random
import matplotlib.pyplot as plt

def approxPi(numSamples):
    count = 0
    for i in range(0, numSamples):
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        if (x*x + y*y < 1):
            count+=1
    return 4*(count/numSamples)

samples = list(range(10, 10001, 20))
absErrors = []
relErrors = []
realPi = math.pi

for i in samples:
    codePi = approxPi(i)
    absError = abs(realPi - codePi)
    absErrors.append(absError)
    relError= abs((realPi - codePi) / realPi)
    relErrors.append(relError)

plt.subplot(2, 1, 1)
plt.plot(samples, absErrors, 'b-')
plt.title('Monte Carlo Pi Absolute Error')
plt.xlabel('Number of Samples')
plt.ylabel('Absolute Error')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(samples, relErrors, 'r-')
plt.title('Monte Carlo Pi Relative Error')
plt.xlabel('Number of Samples')
plt.ylabel('Relative Error')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

def pi_monte_carlo(num_sample_points):
    x = np.random.rand(num_sample_points)
    y = np.random.rand(num_sample_points)

    num_points_inside = (x**2 + y**2) <= 1
    pi_estimate = 4 * np.sum(num_points_inside) / num_sample_points

    return pi_estimate


import matplotlib.pyplot as plt

runs = 100
samples = 1000

absolute_error_vals = []
relative_error_vals = []

for i in range(runs):
    pi_est_value = pi_monte_carlo(samples)

    abs_error = abs(np.pi - pi_est_value)
    rel_error = abs((np.pi - pi_est_value) / np.pi)

    absolute_error_vals.append(abs_error)
    relative_error_vals.append(rel_error)

# Absolute error plot
plt.figure()
plt.plot(range(runs), absolute_error_vals)
plt.xlabel("Run")
plt.ylabel("Absolute Error")
plt.title("Absolute Error over 100 Monte Carlo Runs")
plt.show()

# Relative error plot
plt.figure()
plt.plot(range(runs), relative_error_vals)
plt.xlabel("Run")
plt.ylabel("Relative Error")
plt.title("Relative Error over 100 Monte Carlo Runs")
plt.show()
```
