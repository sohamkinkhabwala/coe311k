```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(None)

plt.figure(figsize=(15, 5))
plt.suptitle("Least Squares Method Parts 1, 2 & 3")

# Part 1 Original Values
plt.subplot(1, 3, 1)

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

A = np.vstack([X, np.ones(len(X))]).T
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]
print(f"Part 1 — Slope: {m}, Intercept: {c}")

x_line = np.linspace(X.min() - 0.5, X.max() + 0.5, 200)
plt.scatter(X, Y, label='Data points')
plt.plot(x_line, m * x_line + c, 'r-', linewidth=2, label=f'Fit: y = {m:.2f}x + {c:.2f}')
plt.title("Part 1")
plt.xlabel("X"); plt.ylabel("Y")
plt.legend()
plt.grid(True)

# Part 2 Randomized Vectors
plt.subplot(1, 3, 2)

X2 = np.random.uniform(0, 10, 20)
Y2 = np.random.uniform(0, 10, 20)

A2 = np.vstack([X2, np.ones(len(X2))]).T
m2, c2 = np.linalg.lstsq(A2, Y2, rcond=None)[0]
print(f"Part 2 — Slope: {m2}, Intercept: {c2}")

x_line2 = np.linspace(X2.min() - 0.5, X2.max() + 0.5, 200)
plt.scatter(X2, Y2, label='Random points')
plt.plot(x_line2, m2 * x_line2 + c2, 'r-', linewidth=2, label=f'Fit: y = {m2:.2f}x + {c2:.2f}')
plt.title("Part 2")
plt.xlabel("X"); plt.ylabel("Y")
plt.legend()
plt.grid(True)

# Part 3 My Linear Equation
plt.subplot(1, 3, 3)

TRUE_SLOPE = 2.5
TRUE_INTERCEPT = 1.0
NOISE_STD = np.random.uniform(0, 1)

x3 = np.arange(0, 1.01, 0.01)
noise = np.random.normal(0, NOISE_STD, len(x3))
y3 = TRUE_SLOPE * x3 + TRUE_INTERCEPT + noise

A3 = np.vstack([x3, np.ones(len(x3))]).T
m3, c3 = np.linalg.lstsq(A3, y3, rcond=None)[0]
print(f"Part 3 — True: y = {TRUE_SLOPE}x + {TRUE_INTERCEPT}  " f"Fitted: y = {m3}x + {c3}")

plt.scatter(x3, y3, s=10, label='Noisy data')
plt.plot(x3, TRUE_SLOPE * x3 + TRUE_INTERCEPT, 'b--', label=f'True: y = {TRUE_SLOPE}x + {TRUE_INTERCEPT}')
plt.plot(x3, m3 * x3 + c3, 'r-', label=f'Fit: y = {m3:.2f}x + {c3:.2f}')
plt.title("Part 3")
plt.xlabel("X"); plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```
