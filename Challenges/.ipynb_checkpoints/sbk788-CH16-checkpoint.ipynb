For k = 1, the solution starts to deviate around h = 0.2, and by h = 0.3 and 0.5 it is unstable. Using a smaller h would prevent this.

For k = 10, the issue is more severe. With h = 0.1 or 0.2, the solution becomes unstable quickly, showing that Euler’s method is not well suited for this case.

```python
import numpy as np
import matplotlib.pyplot as plt

# step sizes
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

# spring constant
k = 1.0

x_start = 0
x_end = 10

# initial conditions
y_0 = 1.0
v_0 = 0.0

# plot position y(x) for each h
plt.figure(figsize=(12, 6))

for h in h_values:
    x = np.arange(x_start, x_end + h, h)
    y = np.zeros(len(x))
    v = np.zeros(len(x))

    # initial conditions
    y[0] = y_0
    v[0] = v_0

    # euler's method loop
    for i in range(1, len(x)):
        y_n = y[i - 1]
        v_n = v[i - 1]

        y[i] = y_n + h * v_n
        v[i] = v_n + h * (-k * y_n)

    plt.plot(x, y, label=f'h = {h}')

plt.title('Position y(x) for different step sizes h (k=1)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()


# k = 10, h = 0.01
k2 = 10
h2 = 0.01

x2 = np.arange(x_start, x_end + h2, h2)
y2 = np.zeros(len(x2))
v2 = np.zeros(len(x2))

y2[0] = y_0
v2[0] = v_0

for i in range(1, len(x2)):
    y_n = y2[i - 1]
    v_n = v2[i - 1]

    y2[i] = y_n + h2 * v_n
    v2[i] = v_n + h2 * (-k2 * y_n)

plt.figure(figsize=(12, 6))
plt.plot(x2, y2, label='position y(x)', color='blue')
plt.plot(x2, v2, label='velocity v(x)', linestyle='--', color='orange')
plt.title('k=10, h=0.01 using euler\'s method')
plt.xlabel('x')
plt.ylabel('y and v')
plt.legend()
plt.grid(True)
plt.show()
```
