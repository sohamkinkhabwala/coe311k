```python
def decimal_to_binary(n):
    if n == 0:
        return "0"

    remainders = ""
    
    while n > 0:
        remainders = str(n % 2) + remainders
        n = n // 2 

    return remainders

num = int(input("Enter a positive integer"))
binaryNum = decimal_to_binary(num)

print(f"Decimal: {num}")
print(f"Binary:  {binaryNum}")
```
