```python
import numpy as np
import time

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations

def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(x_new - x) < tol:
            return x, k
        x = x_new
    return x, max_iterations

def run_challenge(A, b, true_sol, tol, max_iterations):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    x0 = np.zeros(len(b))

    start = time.perf_counter()
    x_jac, iter_jac = jacobi(A, b, x0, tol, max_iterations)
    time_jac = time.perf_counter() - start

    start = time.perf_counter()
    x_gs, iter_gs = gauss_seidel(A, b, x0, tol, max_iterations)
    time_gs = time.perf_counter() - start

    print('Jacobi:')
    print('  Iterations:', iter_jac + 1 if iter_jac < max_iterations else max_iterations)
    print('  Time:', round(time_jac, 6), 's')
    print('  Solution:', np.round(x_jac, 6).tolist())

    print('Gauss-Seidel:')
    print('  Iterations:', iter_gs + 1 if iter_gs < max_iterations else max_iterations)
    print('  Time:', round(time_gs, 6), 's')
    print('  Solution:', np.round(x_gs, 6).tolist())
```

```python
# Challenge 6A
A = [
    [10, -1,  2,  0,  0],
    [-1, 11, -1,  3,  0],
    [ 2, -1, 10, -1,  0],
    [ 0,  3, -1,  8, -2],
    [ 0,  0,  0, -2,  9]
]
b = [14, 30, 26, 25, 37]
true_sol = [1, 2, 3, 4, 5]

print('Challenge 6a')
run_challenge(A, b, true_sol, 10e-6, 100)
```

```python
# Challenge 6b
A = [
    [1, 2, 3, 0, 0],
    [2, 1, 2, 3, 0],
    [3, 2, 1, 2, 3],
    [0, 3, 2, 1, 2],
    [0, 0, 3, 2, 1]
]
b = [14, 22, 33, 26, 22]
true_sol = np.linalg.solve(np.array(A, dtype=float), np.array(b, dtype=float))

print('Challenge 6b')
print('True solution (from numpy):', np.round(true_sol, 6).tolist())
run_challenge(A, b, true_sol, 10e-6, 100)
```
